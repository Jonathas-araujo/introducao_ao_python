#**Tipo Int (Inteiro)**

O tipo de variável `int` é usado principalmente para representar **números inteiros**, ou seja, números que não possuem casas decimais. Exemplos: `2`, `10` e `37`. Ele também pode representar números negativos ou o zero.

> **Nota:** Em Python, basta atribuir um número inteiro a uma variável para que ela seja reconhecida automaticamente como `int`.
>
> ```python
> idade = 20 # O Python já entende que é um int
> ```

A função `int()` também pode ser usada para converter um número decimal em um número inteiro. Porém, isso **não significa arredondar** o número. Por exemplo, `int(7.9)` irá resultar em `7`, pois a função apenas remove a parte decimal.

Em resumo: `int` é o tipo de dado, enquanto `int()` é a função usada para converter um valor para o tipo inteiro.

---

#**Quando usar o `int()`?**

O uso mais comum da função `int()` é quando o valor que temos ainda não é um número inteiro, como acontece com uma *string* (texto) recebida do usuário.

A função `input()` sempre retorna o que foi digitado como texto (`str`). Portanto, mesmo que o usuário digite `20`, o Python inicialmente irá tratá-lo como o texto `"20"`.

Para transformar esse texto em um número com o qual possamos fazer operações matemáticas, usamos o `int()` em volta do `input()`:

**Exemplo de uso:**

````python
idade = int(input("Digite sua idade: "))
print(f"Sua idade é {idade}")
````

# **LIST**

O que é uma Lista (`list`)?

Uma lista (representada no Python pela abreviação `list`) é um tipo de dado estruturado que serve para armazenar vários valores em uma única variável.

Enquanto `int`, `float`, `bool` e `str` guardam apenas um valor por vez, a lista funciona como uma "caixa" ou uma "lista de compras", onde você pode guardar múltiplos itens, sejam eles do mesmo tipo ou misturados (textos, números, booleanos e até outras listas).

Para criar uma lista no Python, usamos colchetes `[]` e separamos os itens por vírgulas.
```python
# Exemplos de listas
notas = [8.5, 9.0, 7.5]                  # Lista de floats
nomes = ["Ana", "Carlos", "Beatriz"]     # Lista de strings
mista = ["Lucas", 25, True, 1.75]        # Lista com vários tipos misturados
```
Como acessar os itens de uma lista? (Indexação)

Assim como vimos nas strings, os itens de uma lista também possuem uma posição numérica chamada índice, e a contagem sempre começa no número 0.


````python
frutas = ["maçã", "banana", "laranja", "uva"]

print(frutas[0])  # Saída: maçã (primeiro item)
print(frutas[1])  # Saída: banana (segundo item)
print(frutas[3])  # Saída: uva (último item)
````

> **Dica útil**
>
> Se você quiser pegar o último item de forma rápida sem saber o tamanho da lista, o Python aceita índices negativos! O índice `[-1]` pega sempre o último elemento, `[-2]` o penúltimo, e assim por diante.

As listas em Python são mutáveis, o que significa que você pode alterar, adicionar ou remover itens depois que a lista já foi criada. (Diferente das strings, que não podem ter letras trocadas individualmente).

1. Alterando um item existente
Você pode mudar o valor de um item específico chamando o seu índice:

````python
cores = ["azul", "verde", "amarelo"]
cores[1] = "vermelho" # Substitui o "verde" por "vermelho"

print(cores)
````

2. Adicionando novos itens (`.append()`)

Para colocar um novo item no final da lista, usamos a ferramenta `.append()`:


````python
tarefas = ["estudar", "limpar a casa"]
tarefas.append("fazer compras")

print(tarefas)
````

3. Removendo itens (`.pop()` ou `del`)

Você pode remover itens indicando a posição deles:

````python
compras = ["pão", "leite", "café"]
compras.pop(0) # Remove o item do índice 0 ("pão")

print(compras)
````

````python
frutas = ["maçã", "banana", "laranja"]

# Remove o item no índice 1 ("banana")
del frutas[1]

print(frutas)
````

Ferramentas Úteis para Listas:

1. Ver o tamanho da lista (`len`): Assim como nas strings, a função `len()` conta quantos itens existem dentro da lista.


````python
alunos = ["João", "Maria", "Pedro"]
print(len(alunos))
````

2. Ordenar a lista (`.sort()`): Organiza os itens em ordem alfabética ou numérica crescente.

````python
numeros = [40, 10, 30, 20]
numeros.sort()
print(numeros)
````

# **DICT**

O que é um Dicionário (`dict`)?

Até agora, vimos diferentes tipos de dados no Python:
* `str` guarda textos.
* `int` e `float` guardam números.
* `list` guarda vários valores organizados por posições numéricas (os índices 0, 1, 2...).

Mas e se quisermos guardar informações onde cada dado tem um nome ou rótulo (em vez de um número de posição)? É exatamente para isso que serve o Dicionário (`dict`).

Enquanto a lista funciona como uma "lista de compras" numerada, o dicionário funciona como um dicionário de verdade: você tem uma Chave (o rótulo) e o seu Valor (a informação).

* **Chave (Key):** É o identificador do dado (ex: `"nome"`, `"idade"`).
* **Valor (Value):** É a informação guardada (ex: `"José"`, `60`).

Para criar um dicionário, usamos chaves `{}` e o sinal de dois-pontos `:` para ligar a chave ao seu valor.

###**Como criar um Dicionário?**

Existem duas formas principais de criar dicionários em Python:

1. **Usando chaves `{}` (A forma mais comum e recomendada)**

```python
# Dicionário de um homem chamado José
homem = {
    "nome": "José",
    "idade": 60,
    "Altura": 170
}
```

2. Usando a função `dict()` (Com o sinal de igual `=`)

Você também pode usar a função `dict()`. Quando usa ela, o Python permite que você escreva o nome da chave sem aspas na hora da criação:

```python
homem = dict(idade=60, nome="José", Altura=170)
```

###**Como acessar os valores de um Dicionário?**

Para buscar um dado específico dentro do dicionário, usamos colchetes `[]` informando o nome da chave.



````python
homem = dict(idade=60, nome="José", Altura=170)

# Buscando a idade
print(homem["idade"])
````


> **Na hora de buscar, a chave PRECISA estar entre aspas!**
>
> Quando criamos o dicionário usando `dict(idade=23...)`, o Python é flexível e aceita que você digite idade sem aspas.
>
> Porém, quando você faz `print(homem["idade"])`, o Python entende que você está fazendo uma busca por texto. Se você esquecer as aspas e tentar `print(homem[idade])`, o Python vai achar que idade é o nome de uma variável solta no seu código e vai dar um erro, dizendo que essa variável não existe.

#**Tuple**
## O que é uma Tupla (`tuple`)?

Até agora, conhecemos estruturas que armazenam vários valores. A **tupla** é muito parecida com uma lista, mas possui uma característica importante: **ela não pode ser alterada depois de criada**.

Em Python:

- `list` → pode ser modificada.
- `tuple` → é **imutável** (não muda).
- Ambas armazenam vários valores em uma sequência ordenada.

As tuplas são muito úteis quando queremos representar informações que devem permanecer fixas, como coordenadas, datas, resultados de exames ou dados de um paciente que não devem ser alterados durante a execução do programa.

> **Exemplo na saúde:** a temperatura corporal medida em um momento específico pode ser armazenada junto com a data em uma tupla, preservando esse registro.

---

## Como criar uma Tupla?

Criamos uma tupla utilizando **parênteses `()`** e separando os valores por vírgulas.

### 1. Criando uma tupla (forma mais comum)

````python
# Temperatura e frequência cardíaca
sinais_vitais = (36.8, 72)

print(sinais_vitais)
````

### 2. Tupla com diferentes tipos de dados

````python
paciente = ("Maria", 45, "O+")
````


Nesse exemplo:

- `"Maria"` → nome
- `45` → idade
- `"O+"` → tipo sanguíneo

---

## Como acessar os valores de uma Tupla?

Assim como nas listas, utilizamos **colchetes `[]`** com o índice da posição.

````python
paciente = ("Maria", 45, "O+")

print(paciente[0])   # Maria
print(paciente[1])   # 45
print(paciente[2])   # O+
````


> Lembre-se: os índices começam em **0**.

---

## Tuplas são imutáveis

Depois de criada, uma tupla **não permite alterar, adicionar ou remover elementos**.

````python
paciente = ("Maria", 45, "O+")

paciente[1] = 46
````

O código acima gera um erro, porque tuplas não podem ser modificadas.

> **Na prática:** use uma **tupla** quando os dados devem permanecer fixos. Se precisar editar os valores, utilize uma **lista**.

# IF, ELIF e ELSE

## O que é o `if`?

A palavra **`if`** significa **"se"**.

Ela é usada para dizer ao Python:

> **Se uma determinada condição for verdadeira, execute um bloco de código abaixo.**

### Exemplo

````python
idade = 18

if idade >= 18:
    print("Você é maior de idade.")
````

### O que acontece aqui?

1. A variável `idade` recebe o valor `18`.
2. O Python verifica a condição `idade >= 18`.
3. Como `18 >= 18` é **verdadeiro**, o `print()` é executado.

**Resultado:**
```python
Você é maior de idade.
```

## Estrutura básica

Todo `if` segue esta estrutura:
```python
if condição:
    código
```

> **Importante:** O símbolo `:` é obrigatório, e todo o código pertencente ao `if` deve estar **indentado** (com espaços ou tabulação).

---

## O que é o `else`?

A palavra **`else`** significa **"senão"**.

Ela é utilizada quando queremos executar um código caso a condição do `if` seja **falsa**.

### Exemplo

````python
idade = 15

if idade >= 18:
    print("Você é maior de idade.")
else:
    print("Você é menor de idade.")
````

### Como o Python pensa?

- A idade é maior ou igual a 18?
- **Não.**
- Então ele executa o bloco do `else`.

**Resultado:**
```python
Você é menor de idade.
```

### Fluxo da decisão
```
          idade >= 18?
           /        \
        SIM          NÃO
         |            |
         ▼            ▼
 Maior de idade  Menor de idade
 ```

---

## O que é o `elif`?

A palavra **`elif`** significa **"senão, se"**.

Ela é usada quando temos **mais de duas possibilidades**.

### Exemplo

````python
nota = 7

if nota >= 9:
    print("Excelente")
elif nota >= 7:
    print("Bom")
else:
    print("Precisa melhorar")
````

### Como o Python verifica?

1. Primeiro ele testa `nota >= 9`.
2. Como `7 >= 9` é **falso**, ele passa para o `elif`.
3. Depois verifica `nota >= 7`.
4. Como `7 >= 7` é **verdadeiro**, ele imprime `"Bom"`.

**Resultado:**
```python
Bom

> O Python verifica as condições **de cima para baixo** e para na primeira condição verdadeira.

---

## Vários `elif`

Podemos utilizar quantos `elif` forem necessários.

### Exemplo

````python
nota = 8

if nota >= 9:
    print("Excelente")
elif nota >= 7:
    print("Bom")
elif nota >= 5:
    print("Regular")
else:
    print("Reprovado")
````

### Lógica de execução
```
nota >= 9?
     │
    Não
     │
nota >= 7?
     │
    Sim
     ▼
    "Bom"
```

Como uma condição foi verdadeira, o Python **não verifica** as próximas.

# FOR e WHILE

## O que é o `for`?

O **`for`** serve para **repetir** uma ação várias vezes.

Imagine que você queira imprimir `"Olá"` cinco vezes. Sem o `for`, precisaríamos escrever o mesmo comando repetidamente.

````python
print("Olá")
print("Olá")
print("Olá")
print("Olá")
print("Olá")
````

Com o `for`, fazemos a mesma coisa de forma muito mais simples:

````python
for i in range(5):
    print("Olá")
````

O Python executará o `print()` **5 vezes**.

---

## O que é esse `i`?

O `i` é uma **variável** que recebe um valor diferente a cada repetição.

````python
for i in range(5):
    print(i)
````

**Resultado:**

````python
0
1
2
3
4
````

A cada repetição acontece o seguinte:

- 1ª vez → `i = 0`
- 2ª vez → `i = 1`
- 3ª vez → `i = 2`
- 4ª vez → `i = 3`
- 5ª vez → `i = 4`

---

## O que é o `range()`?

A função **`range()`** define quantas vezes o `for` será executado.

Por exemplo, `range(5)` representa a sequência de números de **0 até 4**.

> **Importante:** o `range()` começa no **0** e termina **antes** do número informado.

### Exemplo

````python
for i in range(3):
    print(i)
````

**Resultado:**

````python
0
1
2
````

---

## Podemos escolher outro nome?

Sim! O nome da variável pode ser qualquer um.

````python
for numero in range(5):
    print(numero)
````

Também funciona:

````python
for x in range(5):
    print(x)
````

O importante é escolher um nome que faça sentido para o seu código.

---

**Resultado:**

````python
João
Maria
Pedro
````

# O que é o `while`?

O **`while`** significa **"enquanto"**.

Ele repete um bloco de código **enquanto uma condição for verdadeira**.

````python
numero = 1

while numero <= 5:
    print(numero)
    numero += 1
````

**Resultado:**

````python
1
2
3
4
5
````

---

## Como o `while` funciona?

Primeiro criamos uma variável:

````python
numero = 1
````

Depois o `while` verifica continuamente a condição:

````python
while numero <= 5:
    print(numero)
    numero += 1
````

Enquanto `numero <= 5` for verdadeiro, o laço continuará executando.

---

## Cuidado com o loop infinito

Dentro do `while`, a condição precisa mudar. Caso contrário, o programa nunca terminará.

### Correto

````python
numero = 1

while numero <= 5:
    print(numero)
    numero += 1
````

### Incorreto

````python
numero = 1

while numero <= 5:
    print(numero)
````

Nesse exemplo, `numero` permanece igual a `1` para sempre, criando um **loop infinito**.

# `for` × `while`

As duas estruturas servem para repetir código, mas são usadas em situações diferentes.

````python
# FOR: repetir 5 vezes
for i in range(5):
    print("Olá")
````

````python
# WHILE: repetir enquanto a condição for verdadeira
numero = 1

while numero <= 5:
    print(numero)
    numero += 1
````

## Resumo

| Estrutura | Quando usar |
|---|---|
| `for` | Quando sabemos quantas vezes repetir |
| `while` | Quando repetimos enquanto uma condição for verdadeira |

> **Dica:** `for` = **repita X vezes** • `while` = **repita enquanto...**

# DEF — Funções em Python

## O que é uma função?

Uma **função** é um bloco de código que realiza uma tarefa específica.

Ela permite agrupar comandos que podem ser executados sempre que precisarmos, evitando repetir o mesmo código várias vezes.

````python
def saudacao():
    print("Olá!")
````

A palavra **`def`** significa **"definir uma função"**.

Nesse exemplo, estamos criando uma função chamada `saudacao`.

## Executando a função

Depois de criar uma função, basta escrever seu nome seguido de parênteses para executá-la.

````python
saudacao()
````

**Resultado:**

````python
Olá!
````

> **Resumo:**
>
> - `def saudacao():` → cria a função.
> - `saudacao()` → executa a função.

---

## Por que usar funções?

Imagine que você precise mostrar `"Olá!"` três vezes.

### Sem função

````python
print("Olá!")
print("Olá!")
print("Olá!")
````

### Com função

Primeiro criamos a função apenas uma vez:

````python
def saudacao():
    print("Olá!")
````

Depois podemos reutilizá-la quantas vezes quisermos:

````python
saudacao()
saudacao()
saudacao()
````

Assim, escrevemos a lógica uma única vez e reutilizamos sempre que necessário.

---

## Funções que recebem informações

As funções também podem receber valores, chamados de **parâmetros**.

````python
def saudacao(nome):
    print("Olá", nome)
````

Chamando a função:

````python
saudacao("Joao")
````

**Resultado:**

````python
Olá Joao
````

Outro exemplo:

````python
saudacao("Maria")
````

**Resultado:**

````python
Olá Maria
````

### O que é `nome`?

O parâmetro `nome` funciona como uma variável criada dentro da função.

Quando fazemos:

````python
saudacao("João")
````

É como se o Python executasse:

````python
nome = "João"
````

Depois disso, a função utiliza esse valor normalmente.

---

## Funções com números

Uma função também pode receber números e realizar cálculos.

````python
def dobro(numero):
    print(numero * 2)
````

Executando:

````python
dobro(5)
````

**Resultado:**

````python
10
````

Outro exemplo:

````python
dobro(20)
````

**Resultado:**

````python
40
````

A mesma função funciona para qualquer número informado.

---

## Funções com vários parâmetros

Uma função pode receber mais de uma informação.

````python
def soma(a, b):
    print(a + b)
````

Chamando a função:

````python
soma(3, 5)
````

**Resultado:**

````python
8
````

Nesse exemplo:

- `a = 3`
- `b = 5`

O Python calcula `3 + 5` e imprime o resultado.

---

## O que é `return`?

Até agora utilizamos `print()`, que apenas exibe o resultado na tela.

O **`return`** é diferente: ele devolve um valor para que possamos utilizá-lo em outras partes do programa.

````python
def soma2(a, b):
    return a + b
````

Agora podemos guardar o resultado em uma variável:

````python
resultado = soma(3, 5)
````

Depois utilizamos esse valor:

````python
print(resultado)
````

**Resultado:**

````python
8
````

### Como o `return` funciona?
```
soma(3, 5)
     │
     ▼
  3 + 5
     │
     ▼
     8
     │
     ▼
resultado = 8
```

O `return` envia o valor de volta para quem chamou a função.

# `print` × `return`

As duas instruções são importantes, mas têm objetivos diferentes.

### Usando `print`

````python
def soma(a, b):
    print(a + b)
````

O resultado é apenas exibido na tela.

### Usando `return`

````python
def soma(a, b):
    return a + b
````

Agora podemos reutilizar o valor retornado:

````python
resultado = soma(3, 5)
print(resultado)
````

**Resultado:**

````python
8
````

## Resumo

| `print` | `return` |
|---|---|
| Exibe uma informação | Devolve um valor |
| Mostra na tela | Pode ser armazenado em uma variável |
| Usado para visualizar resultados | Usado para reutilizar resultados |

> **Dica:** Use `print()` quando quiser apenas mostrar algo ao usuário. Use `return` quando a função precisar produzir um resultado que será utilizado depois.
