## **1. Importação das Bibliotecas e o Apelido `pd`**

**O que é o Pandas?**
O Pandas é considerado o "Excel do Python". É uma biblioteca construída especificamente para analisar e manipular tabelas de dados.

- **De onde vem o nome?** Vem de *Panel Data* (Dados em Painel), um termo da estatística. Não tem relação com o urso!
- **Por que não usar o Excel?** O Pandas não trava. Ele processa milhões de linhas de pacientes em segundos.
- **Para que serve na prática?** Abrir, limpar e cruzar enormes volumes de dados clínicos para preparar o terreno para a Inteligência Artificial.

**O apelido `pd`:**
Para começar a usar, basta "chamar" essa ferramenta no seu código usando `import pandas as pd`. A sigla "pd" é apenas um apelido carinhoso para não precisarmos digitar a palavra inteira toda vez.

## **2. Series (1D) vs DataFrame (2D) & Criação**

Como o Pandas organiza um sistema de saúde?
* **DataFrame (2D):** É a tabela inteira (linhas e colunas). Imagine a ficha completa com os dados de todos os pacientes.
* **Series (1D):** É uma única coluna. Imagine que você arrancou apenas a coluna "idade" para analisá-la isoladamente.

```python
import pandas as pd
import numpy as np
print("Bibliotecas importadas com sucesso!")

# Criando a tabela exatamente com os dados mostrados nos seus slides
dados = pd.DataFrame({
    "paciente": ["  Ana Lima ", "João", "Maria", "Carlos", "Fernanda", "Lucas", "Beatriz", "João"], # João duplicado para teste
    "idade": ["45", "62", "38", "55", "29", "71", "48", "62"], # Como texto para demonstrar conversão depois
    "glicemia": [99.4582, 140.1234, np.nan, 115.0, -1.0, 158.0, 105.0, 140.1234],
    "sangue": ["A+", "O+", "B+", "AB+", "O-", "A-", "B-", "O+"],
    "cidade": ["Campina Grande", "João Pessoa", "Recife", "Natal", "Fortaleza", "Recife", "Campina Grande", "João Pessoa"],
    "sexo": ["F", "Masc", "Feminino", "M", "F", "M", "Feminino", "Masc"],
    "data_admissao": ["2024-01-15", "2024-02-10", "2024-01-20", "2024-02-10", "2024-03-01", "2024-01-05", "2024-02-28", "2024-02-10"],
    "diagnostico": ["Diabetes Tipo 2", "Hipertensao", "Gripe Forte", "Hipertensao", "Avaliacao", "Diabetes Tipo 1", "Enxaqueca", "Hipertensao"],
    "creatinina": [0.8542, 1.2981, 0.7123, 0.9, 1.0, 1.1, 0.8, 1.2981]
})

# Isolando informações (Os colchetes funcionam sempre. O ponto falha se tiver espaço no nome)
print("--- ISOLANDO UMA SERIES (COLUNA) ---")
display(dados['idade'])
```

## **3. Acessando e Compreendendo a Estrutura (Raio-X)**

Você não quer imprimir 10 mil linhas de uma vez. Para "espiar os prontuários", usamos:
- `.head()`: 5 primeiras linhas.
- `.tail()`: 5 últimas linhas.
- `.sample()`: Sorteia 1 paciente.

Para entender o tamanho e formato da pesquisa clínica:
- `.shape`: Mostra `(linhas, colunas)` (Ex: 5000, 15).
- `.columns`: Mostra o nome de todas as colunas.
- `.info()`: Diz os tipos de dados (texto, número) e se há dados faltantes.
- `.describe()`: Calcula automaticamente a média, mínimo e máximo de todos os números.

```python
print("--- ESPIANDO OS DADOS ---")
display(dados.head()) # 5 primeiras linhas
display(dados.tail()) # 5 últimas linhas
display(dados.sample()) # Sorteia 1 paciente

print("\n--- FORMATO E ESTATÍSTICAS RÁPIDAS ---")
print("Colunas:", dados.columns)
print("Shape:", dados.shape) # Mostra (linhas, colunas)

print("\nInfo:")
dados.info()

print("\nDescribe:")
display(dados.describe())
```

## **4. Valores Únicos, Repetições e o Pesadelo dos Dados Ausentes**

- **Valores Únicos:** `.unique()` lista os valores distintos (ex: Quais os tipos sanguíneos cadastrados?), e `.nunique()` conta quantos valores diferentes existem. `.value_counts()` mostra a frequência (ex: quantos pacientes existem de cada cidade na sua base).
- **Repetições:** Pacientes cadastrados em duplicidade enviesam qualquer modelo de IA. Usamos `.duplicated()` para marcar quais linhas são repetidas e `.drop_duplicates()` para remover e limpar a tabela.
- **Dados Ausentes (NaN):** Na saúde, é normal faltar preencher algum dado. O Pandas marca esses buracos como `np.nan` (Não é um número), e eles quebram a IA. Usamos `.isnull().sum()` para contar, `.replace()` para trocar erros de digitação e `.fillna()` para preencher buracos.

```python
# Valores Únicos
print("Valores únicos de sangue:", dados['sangue'].unique())
print("Quantidade de tipos de sangue:", dados['sangue'].nunique())

print("\nContagem por cidade:")
display(dados['cidade'].value_counts())

# Repetições no sistema
print("\nMarcando linhas repetidas:")
display(dados.duplicated())

# Remove e limpa a tabela
dados = dados.drop_duplicates()

# O pesadelo: dados ausentes
print("\nContando quantos dados estão faltando:")
print(dados.isnull().sum())

# Troca erro de digitação do aparelho (-1.0) por NaN
dados = dados.replace(-1.0, np.nan)

# Preenche o buraco com zero (Conforme slide)
dados = dados.fillna(0)
```

## **5. Conversão de tipos**

* **Conversão de Tipos:** Às vezes, uma idade como 20 é importada como texto. O comando `.astype()` garante a conversão definitiva.

```python
# Igualando as variáveis para rodar o código exato dos slides
df = dados.copy()

# Ajustando a rota (Converte a coluna para números inteiros definitivamente)
df['idade'] = df['idade'].astype(int)
```

## **6. Introdução ao NumPy e Arrays Especiais**

**Afinal, por que importar o NumPy se temos o Pandas?**
Se o Pandas é o organizador de tabelas, o NumPy é o motor matemático dele. Por trás das cortinas, cada coluna do Pandas guarda seus números dentro de um Array do NumPy para conseguir processar os cálculos tão rápido.

**Arrays vs Listas:**
* A **Lista** do Python é como uma maleta médica genérica: cabe gaze, seringa, esparadrapo (aceita texto, número, booleanos tudo junto). É flexível, mas muito lenta para a IA calcular coisas.
* O **Array** do NumPy é como uma bandeja cirúrgica: só aceita um tipo de coisa (ex: apenas números). Essa rigidez permite que o computador processe milhares de operações matemáticas de uma vez só (Vetorização).

**Arrays Especiais:**
- `np.arange()`: Gera uma sequência numérica espaçada, como uma régua.
- `np.zeros()`: Cria matriz preenchida com zeros. Muito usado para inicializar pesos em IA.
- `np.ones()`: Semelhante ao zeros, mas preenche a estrutura com o número 1.

```python
# Criação exata das funções mostradas nos slides de NumPy

# 1. np.arange() - Sequência espaçada
array_regua = np.arange(10) # Gera de 0 a 9
print("np.arange(10):")
print(array_regua)

# 2. np.zeros() - Matriz de zeros
array_zeros = np.zeros(5)
print("\nnp.zeros(5):")
print(array_zeros)

# 3. np.ones() - Matriz de uns
matriz_ones = np.ones(5)
print("\nnp.ones(5):")
print(matriz_ones)
```

Aqui no notebook, usamos o display() apenas para facilitar os testes, permitindo mostrar as duas tabelas dentro de uma única célula de código. Mas ele não é obrigatório: é apenas uma função de exibição, e o código funciona normalmente sem ela. A diferença é só na forma como o resultado aparece na tela.

# **7. Formatando/transformando: .replace(), .round(), pd.to_datetime(), .dt.year, .str.strip(), .str.contains()**





## **.str.strip()**

Quando os dados são digitados manualmente em um sistema, é muito comum que sobrem espaços invisíveis no início ou no final do texto (por exemplo, " Ana Lima " em vez de "Ana Lima"). Isso pode atrapalhar buscas e comparações depois. O método .str.strip() remove esses espaços extras.

```python
import pandas as pd

dados = {
    'paciente': ['  Ana Lima ', 'Carlos Souza   ', ' Beatriz Santos '],
    'convenio': [' Unimed ', ' SUS', 'Bradesco  '],
    'idade': [45, 62, 38]
}
df = pd.DataFrame(dados)

print("ANTES DO STRIP:")
print(df['paciente'].tolist())

# 2. Aplicando a limpeza com .str.strip()
df['paciente'] = df['paciente'].str.strip()
df['convenio'] = df['convenio'].str.strip()

print("\nDEPOIS DO STRIP:")
print(df['paciente'].tolist())
```

## **.replace()**

Em bases clínicas reais, é comum encontrar a mesma informação escrita de formas diferentes ('F', 'Fem', 'Feminino'). O método .replace() permite padronizar esses valores, trocando cada versão pela forma correta.

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos', 'Lucas Rocha'],
    'sexo': ['F', 'Masc', 'Feminino', 'M']
}
df = pd.DataFrame(dados)

# Antes da padronização
print("Antes:")
display(df[['paciente','sexo']])

# Aplicando .replace() para padronizar
df['sexo'] = df['sexo'].replace({'F': 'Feminino', 'M': 'Masculino', 'Masc': 'Masculino'})
print("\nDepois:")
display(df[['paciente','sexo']])

```

## **.round()**

Exames laboratoriais costumam vir com muitas casas decimais, o que dificulta a leitura. O método .round() ajusta o número de casas decimais de uma coluna numérica.

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos'],
    'glicemia': [99.4582, 140.1234, 88.7891],
    'creatinina': [0.8542, 1.2981, 0.7123]
}
df = pd.DataFrame(dados)

print("Antes")
display(df[['paciente', 'glicemia', 'creatinina']])

# Aplicando o .round() para 1 casa decimal na glicemia e 2 na creatinina
df['glicemia'] = df['glicemia'].round(1)
df['creatinina'] = df['creatinina'].round(2)

print("Depois")
display(df[['paciente', 'glicemia', 'creatinina']])
```

## **pd.to_datetime()**

Quando lemos um arquivo (como um CSV), o pandas geralmente interpreta datas como texto simples (tipo object). Isso significa que não conseguimos fazer operações como "quantos dias se passaram" ou "filtrar só os atendimentos de 2024". A função pd.to_datetime() converte esse texto em um formato de data de verdade (datetime64), liberando esses recursos.

> **Nota:** Não é possível fazer operações de data e hora enquanto a coluna estiver como texto (`str`). É por isso que essa conversão é sempre um dos primeiros passos ao tratar dados com datas.

```python
# Base com data de admissão como texto (string)
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos'],
    'data_admissao': ['2023-03-15', '2024-07-20', '2024-09-02']
}
df = pd.DataFrame(dados)

print(df[['data_admissao']].dtypes)

print("Antes:")
display(df[['paciente', 'data_admissao']])

# 2. Convertendo para data real e extraindo apenas o ano
df['data_admissao'] = pd.to_datetime(df['data_admissao'])
df['ano_admissao'] = df['data_admissao'].dt.year

print(df[['data_admissao']].dtypes)
print("Depois:")
display(df[['paciente', 'data_admissao']])
```

## **.dt.year**

Depois que uma coluna está no formato de data, podemos usar o acessor .dt para extrair partes específicas dela, como o ano, o mês ou o dia.

```python
df['ano_admissao'] = df['data_admissao'].dt.year
df[['paciente', 'data_admissao', 'ano_admissao']]
```

> **Dica:** o mesmo acessor funciona para outras partes da data, como .dt.month (mês) e .dt.day (dia).

## **.str.contains()**

Muitas vezes precisamos filtrar linhas que contenham uma palavra específica dentro de um texto — por exemplo, encontrar todos os pacientes cujo diagnóstico menciona "diabetes", mesmo que existam variações de maiúsculas e minúsculas. O parâmetro case=False garante que a busca ignore essa diferença.

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos', 'Lucas Rocha'],
    'diagnostico': ['Diabetes tipo 2', 'Hipertensão Arterial', 'Pré-diabetes', 'Gripe Comum']
}
df = pd.DataFrame(dados)

print("Base Original:")
display(df[['paciente', 'diagnostico']])

# Filtrando pacientes com qualquer menção a 'diabetes' (case=False ignora maiúsculas/minúsculas)
filtro_diabetes = df['diagnostico'].str.contains('diabetes', case=False)
df_diabetes = df[filtro_diabetes]

# Apenas pacientes com diabetes
print("Pacientes Filtrados (com Diabetes):")
display(df_diabetes[['paciente', 'diagnostico']])
```

# **8. Ordenação: .sort_values()**

## **.sort_values()**

Esse método reorganiza as linhas do DataFrame de acordo com os valores de uma coluna. Na prática, isso é muito útil para priorizar atendimentos. Por exemplo, listando primeiro os pacientes com maior glicemia (maior risco).

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos', 'Lucas Rocha'],
    'idade': [45, 62, 38, 71],
    'glicemia': [99.5, 140.1, 88.8, 185.0]
}
df = pd.DataFrame(dados)

print("Antes:")
display(df[['paciente', 'idade', 'glicemia']])

# Ordenando da maior para a menor glicemia (decrescente)
df_ordenado = df.sort_values(by='glicemia', ascending=False)

print("Depois:")
display(df_ordenado[['paciente', 'idade', 'glicemia']])
```

> **Nota:** Para ordenar do menor para o maior, basta usar ascending=True (que já é o padrão, então também funciona sem especificar nada).

# **9. Índice do DataFrame: .set_index(), .reset_index() (incluindo indexar por data como base pra séries temporais)**

## **.set_index() e .reset_index()**

Todo DataFrame tem um índice: os números (0, 1, 2...) que aparecem à esquerda da tabela. Podemos trocar esse índice numérico por uma coluna que identifique melhor cada linha, como o ID do paciente.

```python
dados = {
    'id_paciente': [101, 102, 103],
    'nome': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos'],
    'setor': ['UTI', 'Triagem', 'Enfermaria']
}
df = pd.DataFrame(dados)

# Definindo o ID como índice
df_idx = df.set_index('id_paciente')
print("Com id_paciente no índice (.set_index):")
display(df_idx)

# Resetando para o índice numérico padrão
df_reset = df_idx.reset_index()
print("Índice restaurado (.reset_index):")
display(df_reset)
```

## **Indexação por data**

Quando trabalhamos com dados que se repetem ao longo do tempo (como o número de internações por dia), é muito útil transformar a data em índice. Isso permite consultar um dia específico diretamente com .loc[].

```python
dados_tempo = {
    'data': pd.to_datetime(['2024-05-01', '2024-05-02', '2024-05-03', '2024-05-04']),
    'internacoes_diarias': [12, 18, 15, 22]
}
df_tempo = pd.DataFrame(dados_tempo).set_index('data')

print("Série Temporal (Índice = DatetimeIndex):")
display(df_tempo)
```

```python
print("Internações em 02/05/2024:")
df_tempo.loc['2024-05-02']
```

# **10. Agrupamento e agregação: .groupby()**


## **.groupby()**

O .groupby() agrupa os dados de acordo com uma coluna de categorias (por exemplo, o diagnóstico) e permite calcular resumos estatísticos para cada grupo, como a média, a soma ou a contagem.

```python
dados = {
    'paciente': ['Ana', 'Carlos', 'Beatriz', 'Lucas', 'Marina', 'João'],
    'diagnostico': [
        'Diabetes', 'Hipertensão', 'Diabetes',
        'Gripe', 'Hipertensão', 'Diabetes'
    ],
    'idade': [62, 55, 48, 23, 67, 71]
}

df = pd.DataFrame(dados)

print("Base de Pacientes:")
df[['paciente', 'diagnostico', 'idade']]

# Média de idade por diagnóstico arredondada
resumo = df.groupby('diagnostico')[['idade']].mean().round(1)

print("Média de Idade por Diagnóstico:")
resumo.reset_index()
```

> **Nota:** Outras agregações úteis: .sum() (soma), .count() (contagem de registros) e .max() / .min() (maior e menor valor).

# **11. Criando novas variáveis: coluna condicional com .loc[], coluna calculada com .assign()**


## **.loc[]**

Podemos criar uma nova coluna cujo valor depende de uma regra lógica. Um exemplo clássico na saúde é classificar a glicemia como "Normal" ou "Alterada" de acordo com um valor de referência clínico (neste caso, 126 mg/dL, o limiar usado para diagnóstico de diabetes em jejum).

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos', 'Lucas Rocha'],
    'glicemia': [95.0, 145.0, 88.0, 180.0]
}
df = pd.DataFrame(dados)

print("Base Original:")
display(df[['paciente', 'glicemia']])

# Criando coluna condicional com .loc[]
df['status'] = 'Normal'
df.loc[df['glicemia'] >= 126.0, 'status'] = 'Alterada'

print("Com Coluna Condicional:")
display(df[['paciente', 'glicemia', 'status']])
```

## **.assign()**

O .assign() permite criar novas colunas a partir de uma função aplicada sobre colunas já existentes, de forma organizada e sem precisar reescrever o DataFrame inteiro. Um exemplo prático é calcular o IMC a partir do peso e da altura.

```python
dados_imc = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos'],
    'peso': [68.0, 85.0, 54.0],
    'altura': [1.65, 1.75, 1.60]
}
df_imc = pd.DataFrame(dados_imc)

print("Antes do Cálculo:")
display(df_imc[['paciente', 'peso', 'altura']])

# Definindo a função clínica para cálculo do IMC
def calcular_imc(tabela):
    return (tabela['peso'] / (tabela['altura'] ** 2)).round(1)

# Aplicando com .assign()
df_imc = df_imc.assign(imc=calcular_imc)

print("Depois (.assign):")
display(df_imc[['paciente', 'peso', 'altura', 'imc']])
```

# **12. Removendo colunas: dados.drop('COLUNA', axis=1)**


## **.drop()**

Às vezes precisamos eliminar colunas desnecessárias ou que contenham informações sensíveis antes de compartilhar uma base de dados. Usamos o parâmetro axis=1 para indicar que queremos remover uma coluna (e não uma linha).

```python
dados = {
    'paciente': ['Ana Lima', 'Carlos Souza', 'Beatriz Santos'],
    'diagnostico': ['Diabetes', 'Hipertensão', 'Diabetes'],
    'observacoes_internas': ['Anotação 1', 'Anotação 2', 'Anotação 3']
}
df = pd.DataFrame(dados)

print("Base Original (com observacoes_internas):")
display(df[['paciente', 'diagnostico', 'observacoes_internas']])

# Removendo a coluna desnecessária
df_limpo = df.drop('observacoes_internas', axis=1)

print("Base Limpa (sem a coluna):")
display(df_limpo)
```
